Our project aims to recommend jobs based on the user’s skills,

- show missing skills,
- give a short learning plan.

We test two prompt templates and compare their results to choose the best one.


## 1. Setup & Imports

```python
import os
from openai import OpenAI
import pandas as pd
from dotenv import load_dotenv

# Load API Key from the .env file
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")

# Check if key is loaded
if not api_key:
    raise ValueError("API Key not found! Please check your .env file.")

# Initialize OpenAI Client
client = OpenAI(api_key=api_key)
print("Setup Complete. API Key loaded successfully.")

# Helper function to call the API
def call_gpt(prompt, model="gpt-4o-mini"):
    try:
        response = client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}],
            temperature=0.7
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"Error: {e}"
```

# 2. Example user profile for testing prompts

```python
user_profile = {
    "name": "Sara",
    "skills": ["Python", "SQL", "Excel", "data visualization", "problem solving"],
    "career_interest": "Data Analytics",
    "experience_level": "Beginner"
}
```

## Template 1 – Policy-Based Career Advisor

This template takes the user profile as input and produces structured career advice.

It focuses on:
- explaining why the job fits the user,
- identifying missing skills,
- providing a short learning plan,
- giving final professional advice.

```python
template1 = """
You are an AI career advisor agent.

The environment is a career recommendation system.
The current state of the user is:

- Skills: {skills}
- Predicted job title: {job_title}
- Category: {category}
- Cluster group: {cluster}

Your task is to act as a policy that generates the best possible action for this state.

Generate a career advice report with the following structure:

1. Career Matching Explanation
2. Missing Skills
3. Learning Plan (3-5 steps)
4. Final Advice

Focus on clarity, relevance, and usefulness.
"""
```

## Template 2 – Conversational Mentor Approach

This template uses a more conversational, mentor-like style to provide career advice.

It focuses on:
- building a personal connection with the user,
- using storytelling and examples,
- providing empathetic guidance,
- offering actionable next steps in a friendly tone.


```python
template2 = """
You are a friendly and experienced career mentor having a conversation with a job seeker.

The user's profile:
- Name: {name}
- Current Skills: {skills}
- Career Interest: {career_interest}
- Experience Level: {experience_level}
- Recommended Job Title: {job_title}
- Job Category: {category}
- Job Cluster: {cluster}

As their mentor, provide personalized career guidance in a warm, conversational tone. 

Your response should:
1. Start with an encouraging greeting that acknowledges their skills and goals
2. Explain why this career path is a good fit for them (use specific examples from their skills)
3. Identify 2-3 key skills they should develop next (explain why each matters)
4. Share a practical learning roadmap with concrete steps they can take this week/month
5. End with motivational advice and next steps

Write as if you're having a one-on-one conversation. Be encouraging, specific, and actionable.
"""
```

## 3. Testing Template 1


```python
# Sample job recommendation data (simulating Phase 2 & 3 outputs)
sample_job_data = {
    "job_title": "Data Analyst",
    "category": "Analytics",
    "cluster": "Data Science & Analytics"
}

# Format Template 1 prompt
prompt1 = template1.format(
    skills=", ".join(user_profile["skills"]),
    job_title=sample_job_data["job_title"],
    category=sample_job_data["category"],
    cluster=sample_job_data["cluster"]
)

print("=" * 80)
print("TEMPLATE 1 OUTPUT:")
print("=" * 80)
response1 = call_gpt(prompt1)
print(response1)
print("\n")
```

## 4. Testing Template 2


```python
# Format Template 2 prompt
prompt2 = template2.format(
    name=user_profile["name"],
    skills=", ".join(user_profile["skills"]),
    career_interest=user_profile["career_interest"],
    experience_level=user_profile["experience_level"],
    job_title=sample_job_data["job_title"],
    category=sample_job_data["category"],
    cluster=sample_job_data["cluster"]
)

print("=" * 80)
print("TEMPLATE 2 OUTPUT:")
print("=" * 80)
response2 = call_gpt(prompt2)
print(response2)
print("\n")
```

### Template Comparison
To evaluate the performance of the two prompt templates, a structured comparison was conducted using six key aspects: tone, structure, personalization, actionability, user engagement, and level of detail. The following analysis is based directly on the comparison table generated in the notebook.

1. Tone & Style

The two templates differ significantly in tone.

Template 1 adopts a formal and agent-like tone suitable for a professional or policy-driven application.

Template 2 uses a warm and friendly mentor-like style, making the output more approachable and emotionally supportive.

2. Structure

Template 1 follows a strict four-section format, ensuring clarity, consistency, and predictability across outputs.

Template 2 is more flexible and conversational, which allows for natural flow but may introduce variability.

3. Personalization

Template 1 provides moderate personalization by using user data but keeps the tone neutral.

Template 2 offers high personalization, directly addressing the user with a more human-like and encouraging approach.

4. Actionability

Both templates perform well in terms of actionability.

Template 1 provides a clear and structured learning plan, making it highly actionable.

Template 2 provides a practical and relatable roadmap, also maintaining strong actionability but in a conversational format.

5. User Engagement

Template 1 is professional but less engaging due to its formal tone.

Template 2 delivers high engagement, using encouraging language that feels more supportive to users.

6. Detail Level

Template 1 offers comprehensive and highly detailed information.

Template 2 also provides detailed outputs but enriches them with examples, making the information easier to visualize and understand.

### Conclusion of Comparison

Both templates provide strong outputs but serve different user experiences:

Template 1 is best suited for formal, structured, and highly consistent system responses.

Template 2 excels in personalization, engagement, and a friendly user experience.

our choice between them is Template 2 (Conversational Mentor Approach) because it delivers a superior user experience.

It uses a warm, friendly, and engaging tone.

It provides higher personalization (using the user's name and experience level), which makes the advice feel more relevant.

It creates a stronger emotional connection, making the career guidance more memorable and impactful for the end-user.

In short, it is better for giving personalized, motivating advice to real people than the formal, agent-like Template 1.

### Justification for Final Template Selection
The most suitable template for a user-facing career recommendation system is Template 2 (Conversational Mentor Approach).

The primary objective of this phase is to make the automated outputs from Phase 2 and 3 useful for real users. This means the advice must not only be technically accurate but also highly engaging, personalized, and motivating.

Key Reasons for Selecting Template 2:
Superior User Engagement and Experience: Template 2's warm, friendly, and conversational tone creates an emotional connection, making the advice more memorable and impactful for the end-user. Template 1, while professional, risks feeling impersonal or robotic.

Higher Relevance and Personalization: Template 2 specifically incorporates the user's name, career interest, and experience level to tailor the advice (e.g., beginner-friendly advice for a "Beginner" user). This makes the advice feel highly relevant (Score: 10/10) compared to Template 1's more generic, though relevant, approach (Score: 9/10).

Actionable and Holistic Guidance: While both templates offer high actionability, Template 2 is designed to start with an encouraging greeting and end with motivational advice, framing the learning plan within a supportive context.

In summary, for a project designed to offer a valuable and human-centered user experience for career guidance, Template 2 is the optimal choice because its focus on personal connection and high engagement directly supports the goal of providing clear, meaningful career advice. Template 1 is better suited for purely automated or formal reporting systems where human readability is a secondary concern.
